# Task

Estimate a market-compliant price of a diamond with a length ($x$) of 16mm, a width ($y$) of 16mm, and a depth $z$ of 10mm. You have a collection of data of prices obtained so far at your disposal to create a model.

```
import pandas as pd

df = pd.read_csv("DataSets/diamonds.csv")
```

```
df.head()
```

```
X = df[['x', 'y', 'z']]
y = df['price']


from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)


# নতুন হীরার মাপ একটি ডাটাফ্রেম বা থ্রি-ডি অ্যারে আকারে দিতে হবে
new_diamond = [[16, 16, 10]]
predicted_price = model.predict(new_diamond)

print(f"আনুমানিক বাজার মূল্য: {predicted_price[0]}")

```
