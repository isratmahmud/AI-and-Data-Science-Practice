```
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (12, 5)

from sklearn.linear_model import LinearRegression
```

```
m=5
n=3

x = np.linspace(0,100).reshape(-1,1)
y_org = m*x+n
```

```
mu = 0;
sigma = 100;

y = y_org + sigma * np.random.randn(y_org.shape[0],1) + mu
```

# Task

Given input data generated by a linear function $y=m \cdot x+n$ and subsequently noisy with a Gaussian distributed sample. Estimate a model on the noisy data using linear regression. Compare the coefficient of determination of the model estimate with respect to the training data $y$ and the actual underlying function $y\_org$.
Visualize the data and the model estimate. Then investigate the influence of the signal-to-noise ratio. What happens if you increase the noise?

```
# ৪. লিনিয়ার রিগ্রেশন মডেল তৈরি এবং ফিট করা
model = LinearRegression()
model.fit(x, y)

# মডেল থেকে পাওয়া প্রেডিকশন (Prediction)
y_pred = model.predict(x)
```

```
# ৫. Coefficient of Determination (R² Score) হিসাব করা
from sklearn.metrics import r2_score

r2_noisy = model.score(x, y)          # নয়েজি ডাটার সাপেক্ষে R² স্কোর
r2_actual = r2_score(y_org, y_pred)   # আসল ফাংশনের সাপেক্ষে R² স্কোর

print(f"Coefficient of Determination (R²):")
print(f"- নয়েজি ডাটার ($y$) সাপেক্ষে: {r2_noisy:.4f}")
print(f"- আসল ফাংশনের ($y_org$) সাপেক্ষে: {r2_actual:.4f}")
print(f"- মডেলের আনুমানিক ঢাল (Estimated Slope): {model.coef_[0][0]:.4f} (আসল ছিল: {m})")
print(f"- মডেলের আনুমানিক ইন্টারসেপ্ট (Estimated Intercept): {model.intercept_[0]:.4f} (আসল ছিল: {n})")

# ৬. ভিজ্যুয়ালাইজেশন (Data Visualization)
plt.scatter(x, y, color='gray', alpha=0.6, label='Noisy Data (y)')
plt.plot(x, y_org, color='blue', linewidth=2, label='Original Line (y_org)')
plt.plot(x, y_pred, color='red', linestyle='--', linewidth=2, label='Model Estimate (y_pred)')
plt.title(f'Linear Regression with Noise (Sigma = {sigma})')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```
