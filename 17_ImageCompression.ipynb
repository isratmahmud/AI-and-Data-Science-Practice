## Task

Cluster the colors in the RGB image 'lena_color.gif'. Use the found prototypes as indexed color palette. The function `showImage()` defined in the following takes over the visualization. The call looks like this:

    showImage(labels, prototypes, width=512, height=512)
    
- `labels` .. cluster membership of each pixel
- `prototypes` .. The color prototypes
- `width`& `height` ... width and height of the clustered image (default values refer to `lena_color.gif` .

Experiment with the number of clusters and use other images as well. What do you observe?

```
import matplotlib.image as mpimg
import numpy as np

img = mpimg.imread("lena_color.gif")
```

```
import matplotlib.pyplot as plt
import matplotlib.colors as col

def showImage(labels, prototypes, width=512, height=512):
    ### 
    cm=col.ListedColormap(prototypes/255.0)
    plt.figure(figsize=(20,20))
    plt.imshow(labels.reshape(512,512),cmap=cm)
    matfig=plt.figure(figsize=(20,2))
    plt.matshow(np.arange(0,prototypes.shape[0]).reshape(1,-1), cmap=cm, fignum=matfig.number)
```

```
plt.imshow(img)
```

```
img.shape
```

```
from sklearn.cluster import KMeans
cluster_model = KMeans(n_clusters=100, verbose=False, n_init=2)
cluster_model.fit(img.reshape(-1,4))
```

```
showImage(cluster_model.labels_, cluster_model.cluster_centers_)
```

```
from sklearn.cluster import KMeans
cluster_model = KMeans(n_clusters=1000, verbose=False, n_init=2)
cluster_model.fit(img.reshape(-1,4))

showImage(cluster_model.labels_, cluster_model.cluster_centers_)
```
